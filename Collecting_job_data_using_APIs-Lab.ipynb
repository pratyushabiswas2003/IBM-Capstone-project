```python
import requests # you need this module to make an API call
import pandas as pd
```

```python
api_url = "http://api.open-notify.org/astros.json" # this url gives use the astronaut data
```

```python
response = requests.get(api_url) # Call the API using the get method and store the
                                # output of the API call in a variable called response.
```

```python
if response.ok:             # if all is well() no errors, no network timeouts)
    data = response.json()  # store the result in json format in a variable called data
                            # the variable data is of type dictionary.
```

```python
print(data)   # print the data just to check the output or for debugging
```

Print the number of astronauts currently on ISS.


```python
print(data.get('number'))
```

Print the names of the astronauts currently on ISS.


```python
astronauts = data.get('people')
print("There are {} astronauts on ISS".format(len(astronauts)))
print("And their names are :")
for astronaut in astronauts:
    print(astronaut.get('name'))
```

```python
api_url="https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/module%201/Accessing%20Data%20Using%20APIs/jobs.json"

def get_number_of_jobs_T(technology):
    #your code goes here
    response = requests.get(api_url)
    if response.ok:
        data = response.json()
        number_of_jobs = sum(1 for job in data if technology.lower() in job["Key Skills"].lower())
        return technology, number_of_jobs
    else:
        return technology, 0
    return technology,number_of_jobs
```

#### Write a function to find number of jobs in US for a location of your choice


```python
 def get_number_of_jobs_L(location):
    
    #your coe goes here
    response = requests.get(api_url)
    if response.ok:
        data = response.json()
        number_of_jobs = sum(1 for job in data if location.lower() in job["Location"].lower())
        return location, number_of_jobs
    else:
        return location, 0
    return location,number_of_jobs
```

Create a python list of all technologies for which you need to find the number of jobs postings.


```python
#your code goes here
locations = ["Los Angeles", "New York", "San Francisco", "Washington DC", "Seattle", "Austin", "Detroit"]

location_results = [get_number_of_jobs_L(loc) for loc in locations]
```

Import libraries required to create excel spreadsheet


```python
# your code goes here
#Import required libraries
import pandas as pd
import json
import requests
from openpyxl import Workbook
```

Create a workbook and select the active worksheet


```python
# your code goes here
wb = Workbook()
ws = wb.active
ws.title = "Job Postings by Location"

# Write headers
ws.append(["Location", "Number of Job Postings"])

# Write data
for loc, count in location_results:
    ws.append([loc, count])
```

Find the number of jobs postings for each of the technology in the above list.
Write the technology name and the number of jobs postings into the excel spreadsheet.


```python
#your code goes here
technologies = ["C", "C#", "C++", "Java", "JavaScript", "Python",
                "Scala", "Oracle", "SQL Server", "MySQL Server",
                "PostgreSQL", "MongoDB"]

tech_results = [get_number_of_jobs_T(tech) for tech in technologies]
tech_results
```

Save into an excel spreadsheet named **job-postings.xlsx**.


```python
#your code goes here
ws2 = wb.create_sheet(title="Job Postings by Technology")
ws2.append(["Technology", "Number of Job Postings"])
for tech, count in tech_results:
    ws2.append([tech, count])

# Save the file
wb.save("job-postings.xlsx")
print("Saved job-postings.xlsx with two sheets: locations and technologies.")

```
